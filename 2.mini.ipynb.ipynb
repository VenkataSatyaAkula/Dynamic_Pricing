```python
x = df[
    [
        "product_id",
        "product_category_name",
        "qty",
        "freight_price",
        "product_photos_qty",
        "lag_price"
    ]
]

y = df["unit_price"]
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42
)
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
print("Training Data Size:", len(x_train))
print("Testing Data Size:", len(x_test))
```

```python
selected_features=df[["product_id","product_category_name","qty","freight_price","product_photos_qty","lag_price"]]
selected_features
```

```python
from sklearn.linear_model import LinearRegression
linearRegression=LinearRegression()
linearRegression.fit(x_train,y_train)
print("model trained successfully")
```

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    root_mean_squared_error,
    r2_score
)
```

```python
y_train_pred = linearRegression.predict(x_train)
train_comparison = pd.DataFrame({'Actual_Train': y_train.values.ravel(), 'Predicted_Train': y_train_pred})
display(train_comparison.head(10))
```

```python
y_pred=linearRegression.predict(x_test)
comparison=pd.DataFrame({'Actual':y_test.values.ravel(),'Predicted':y_pred})
comparison.head(10)
```

```python
from sklearn.metrics import mean_squared_error
mse=mean_squared_error(y_test,y_pred)
print(mse)
```

```python
from sklearn.metrics import mean_absolute_error
mae=mean_absolute_error(y_test,y_pred)
print(mae)
```

```python
from sklearn import metrics
print('R2_score',metrics.r2_score(y_test,y_pred))
```
